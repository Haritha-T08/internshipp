```python
import pandas as pd
```

```python
#	Create a series score from the list of scores with student names as the index. (3)
score_series=pd.Series([78,85,92,67,74,88,95,61],index=["Anu","Binu","Cathy","Deepa","Ebin","Fathima","Gopal","Hari"])
print(score_series)
```

```python
#	Create a series dept from a Python dictionary mapping student to department. (3)
dept_series=pd.Series({"Anu":"CS","Binu":"IT","Cathy":"CS","Deepa":"BCA","Ebin":"IT","Fathima":"CS","Gopal":"CS","Hari":"BCA"})
print(dept_series)
```

```python
#Set the name of scores to "Test Score" and display it. (2)
score_series.name="Test score"
print("Set the name of scores to Test Score and display it\n",score_series.name)
```

```python
#Display the values index data type and name of score using the attributes
print("values of score using the values attribute",score_series.values)

print("index of score using the index attribute",score_series.index)

print("data type  of score using the dtype attribute",score_series.dtype)

print("name of score using the name attribute",score_series.name)
```

```python
#Display Cathy's score using its label. (2)
print("Cathy's score using its label\n",score_series.loc["Cathy"])
```

```python
#Display the scores of Anu, Cathy and Ebin using multi-label indexing. (3)
print("scores of Anu, Cathy and Ebin using multi-label indexing\n",score_series.loc[["Anu","Cathy","Ebin"]])
```

```python
#	Find the highest score, the lowest score, the mean (rounded to 2 decimals) and the number of students. (4)
print("Highest score=",score_series.max())

print("lowest score=",score_series.min())

print("mean of the total score=",score_series.mean())

print("number of students=",score_series.count())

```

```python
#	Display the students who scored 80 or above. (3)
print("students who scored 80 or above\n",score_series[score_series>=80])
```

```python
#Display the top 3 scores in descending order. (3)
n=score_series.sort_values()
print("top 3 scores in descending order\n",n[-3:])
```

```python
#	Find the name of the student with the highest score. (2)
n=score_series.sort_values()
print("name of the student with the highest score\n",n[-1:])
```

```python
#	Write a function grade(score) that returns:
#a.	90 and above: "Distinction"
#b.	75 to 89: "First Class"
#c.	60 to 74: "Second Class"
#d.	below 60: "Fail"
#Apply it to scores and store the result in grades. (5)

def grade(score):
  if score>=90:
    return "Distinction"
  elif 75<=score<89:
    return "First class"
  elif 60<=score<=74:
    return "Second class"
  else:
    return "Fail"
result=score_series.apply(grade)
print(result)

```

```python
#	Display how many students fall in each grade, then the same as percentages. (4)

print(result.value_counts(normalize=True)*100)
```

```python
#Display the unique grades and the number of unique grades. (2)
print(" unique grades and the number of unique grades\n",result.unique())
```
